### Applying Binning (Discretization)

```python
age_col_binning = KBinsDiscretizer(n_bins=15,strategy='quantile',encode='ordinal')
fare_col_binning = KBinsDiscretizer(n_bins=15,strategy='uniform',encode='ordinal')
```

```python
trf = ColumnTransformer([
    ('trf1', age_col_binning, [0]),
    ('trf2', fare_col_binning, [1])
])
```

```python
trf_x_train_trans = trf.fit_transform (X_train)
trf_x_test_trans = trf.transform(X_test)
```

```python
df.dropna(inplace=True)
```

```python
trf.named_transformers_.trf1.bin_edges_
```

```python
clf.fit(trf_x_train_trans,y_train)
```

```python
y_pred1 = clf.predict(trf_x_test_trans)
```

```python
accuracy_score(y_test,y_pred1)
```

```python
np.mean(cross_val_score(clf,df.iloc[:,1:],df['Survived'],scoring='accuracy',cv=10))
```

```python
def binning (method, n):
    age_binning = KBinsDiscretizer(strategy = method, n_bins=n, encode='ordinal')
    fare_binning = KBinsDiscretizer(strategy = method, n_bins=n, encode='ordinal')

    trf=ColumnTransformer([
        ('trf1', age_binning, [0]),
        ('trf2', fare_binning, [1])
    ])

    X_trf = trf.fit_transform(df.iloc[:,1:])
    print(np.mean(cross_val_score(DecisionTreeClassifier(),df.iloc[:,1:],df['Survived'],cv=10,scoring='accuracy')))
    
    plt.figure(figsize=(14,4))
    plt.subplot(121)
    plt.hist(df['Age'])
    plt.title("Before")

    plt.subplot(122)
    plt.hist(X_trf[:,0],color='red')
    plt.title("After")

    plt.show()
    
    plt.figure(figsize=(14,4))
    plt.subplot(121)
    plt.hist(df['Fare'])
    plt.title("Before")

    plt.subplot(122)
    plt.hist(X_trf[:,1],color='red')
    plt.title("Fare")

    plt.show()
    
```

```python
binning ('quantile', 10)
```

```python
binning ('kmeans', 10)
```

```python
binning('uniform',15)
```

### Binarization --> a special type of binning in which we take a certain threshold and divide the bins into 1 and 0 (only two)

```python
from sklearn.preprocessing import Binarizer
```

```python
df_1 = pd.read_csv(Path.home()/"AI Track/Machine-Learning/Binning and BInarization/Datasets/train.csv",usecols=['Survived','Age','Fare','SibSp','Parch'])
```

```python
df_1
```

```python
df_1['family'] = df_1['SibSp'] + df_1['Parch']
df_1.sample(5)
```

```python
df_1.dropna(inplace=True)
```

```python
X_tr, X_te, y_tr, y_te = train_test_split (df_1.iloc[:,1:],df_1['Survived'],test_size=0.3,random_state=42)
```

```python
clf.predict(X_te)
```
