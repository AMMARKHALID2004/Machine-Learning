#### The number column has both numerical and categorical data
I will make two new columns from it

```python
df['numerical_number'] = pd.to_numeric(df['number'], errors = 'coerce', downcast = 'integer' )
df['categorical_number'] = np.where(df['numerical_number'].isna(),df['number'],np.nan)
```

```python
df.isna().sum()
```

### Ticket column has both Categorical and Numerical data together in single cells
lets make new columns for this too

```python
df['Ticket_Category'] = df['Ticket'].str.split().str.get(0)
df['Ticket_Number'] = df['Ticket'].str.split().str.get(1)
```
